STEMMING.

```python
words = ["game","gaming","gamed","games","writing","writes","eating","eat"]
#
```

```python
from nltk.stem import PorterStemmer

stemming = PorterStemmer()
for word in words:
  print(word+"----->"+stemming.stem(word))
```

REGEX STEMMER CLASS

```python
from nltk.stem import RegexpStemmer
stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4) # $ means only the last one
stemmer.stem("ingeating")
```

```python
from nltk.stem import RegexpStemmer
stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4) # $ means only the last one, min of 4 characters
stemmer.stem("unable")
```
