LEMMATIZATION

```python
import nltk
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
#### POS noun - n , verb - v , adjective - a , adverb -r
print(lemmatizer.lemmatize("going",pos='n'))
print(lemmatizer.lemmatize("going",pos='v'))
print(lemmatizer.lemmatize("going",pos='a'))
print(lemmatizer.lemmatize("going",pos='r'))
```

```python
words = ["game","gaming","gamed","games","writing","writes","eating","eat","history","programs"]
```

```python
for word in words:
  print(word+"----->"+lemmatizer.lemmatize(word,pos='v'))
```
